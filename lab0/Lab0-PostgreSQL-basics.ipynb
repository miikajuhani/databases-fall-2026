## 3. Inspect where you are

A PostgreSQL server can contain several databases. Tables normally live in schemas; `public` is the default schema for this course.

```python
%%sql
SELECT current_database() AS database_name,
       current_user AS connected_as,
       current_schema() AS schema_name;
```

```python
%%sql
SELECT table_schema, table_name
FROM information_schema.tables
WHERE table_type = 'BASE TABLE'
  AND table_schema NOT IN ('pg_catalog', 'information_schema')
ORDER BY table_schema, table_name;
```

## 4. Reset and create the practice table

This is the reset cell. It removes only `lab0_books`, then creates an empty table again. The table has a primary key, required values (`NOT NULL`), a default value, and a `CHECK` constraint. You may run this cell repeatedly.

```python
%%sql
DROP TABLE IF EXISTS lab0_books;

CREATE TABLE lab0_books (
    book_id INTEGER PRIMARY KEY,
    title TEXT NOT NULL,
    author TEXT NOT NULL,
    published_year INTEGER CHECK (published_year BETWEEN 1450 AND 2100),
    in_stock BOOLEAN NOT NULL DEFAULT TRUE
);
```

## 5. Insert a small dataset

A single `INSERT` statement can add several rows. Run this cell once after the reset cell.

```python
%%sql
INSERT INTO lab0_books (book_id, title, author, published_year, in_stock) VALUES
    (1, 'The Left Hand of Darkness', 'Ursula K. Le Guin', 1969, TRUE),
    (2, 'The Dispossessed', 'Ursula K. Le Guin', 1974, TRUE),
    (3, 'Never Let Me Go', 'Kazuo Ishiguro', 2005, FALSE),
    (4, 'Piranesi', 'Susanna Clarke', 2020, TRUE),
    (5, 'Kindred', 'Octavia E. Butler', 1979, FALSE);
```

## 6. Read data with `SELECT`

Start by reading every column and every row. Then use `WHERE`, `ORDER BY`, and `LIMIT` to ask a more focused question.

```python
%%sql
SELECT *
FROM lab0_books;
```

```python
%%sql
SELECT title, author, published_year
FROM lab0_books
WHERE in_stock = TRUE
ORDER BY published_year DESC
LIMIT 3;
```

## 7. Change and remove data

`UPDATE` changes existing rows. `DELETE` removes matching rows. The final `SELECT` lets you confirm both changes.

```python
%%sql
UPDATE lab0_books
SET in_stock = TRUE
WHERE book_id = 3;
```

```python
%%sql
DELETE FROM lab0_books
WHERE book_id = 5;
```

```python
%%sql
SELECT book_id, title, in_stock
FROM lab0_books
ORDER BY book_id;
```
